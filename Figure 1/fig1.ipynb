```python
import pandas as pd 
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
import seaborn as sns
from scipy.signal import butter,filtfilt
from scipy.interpolate import make_interp_spline, BSpline
from scipy.stats import pearsonr, linregress
from adjustText import adjust_text
import statsmodels.api as sm
from matplotlib.colors import LinearSegmentedColormap
matplotlib.rc('xtick', labelsize=12) 
matplotlib.rc('ytick', labelsize=12) 
```

```python
# Load predictions and merge
df1 = pd.read_excel("dpsdi_stacked_roc_oot.xlsx")
df2 = pd.read_excel("dpsdi_stacked_roc_oocv1.xlsx")
df3 = pd.read_excel("dpsdi_stacked_roc_oocv2.xlsx")

# Plot the ROC curve
plt.figure(figsize=(5, 5))  
plt.plot(df1['FPR'].values, df1['TPR'].values, color='tab:blue', lw=4, alpha = 0.75, label = r'±DEP PPND SSD (OOT)')
plt.plot(df2['FPR'].values, df2['TPR'].values, color='tab:orange', lw=4, alpha = 0.75, label = r'±DEP Psy-ShareD SSD (OOCV)')
plt.plot(df3['FPR'].values, df3['TPR'].values, color='tab:red', lw=4, alpha = 0.75, label = r'MDD v HC, PPND (OOCV)')

# Plot the diagonal line
plt.plot([0, 1], [0, 1], color='grey', lw=2, linestyle='--')

# Add labels and title
plt.xlabel('False Positive Rate (FPR)', fontsize=14)
plt.ylabel('True Positive Rate (TPR)', fontsize=14)
#plt.title('ROC Curve 6 month ±ΔCDSS classifiers', fontsize=12)

plt.xlim([0,1])
plt.ylim([0,1])

plt.legend(loc='lower right')

# Add grid and ticks
plt.grid(True)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# Ensure the aspect ratio is equal
plt.gca().set_aspect('equal', adjustable='box')

plt.savefig('fig1.1.pdf', format='pdf', bbox_inches='tight')
# Show the plot
plt.show()
```
